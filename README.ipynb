# Análise de NPS em E-commerce

## Objetivo do projeto

O objetivo deste projeto é entender os principais fatores que impactam a satisfação dos clientes (NPS) em um e-commerce, identificando pontos de melhoria na experiência de compra.

---

## Descrição da base de dados

A base contém informações de pedidos realizados por clientes, incluindo:

- Nota de NPS atribuída após a compra
- Tempo de entrega
- Atraso na entrega (em dias)
- Região do cliente
- Número de contatos com o atendimento

Cada linha representa a avaliação de um cliente após a experiência de compra.

---

## Metodologia utilizada

A análise foi conduzida em etapas:

1. *Tratamento dos dados*
   - Substituição de valores inválidos ("-")
   - Conversão de tipos de dados
   - Remoção de registros sem NPS

2. *Classificação do NPS*
   - Detratores: notas abaixo de 7  
   - Neutros: notas entre 7 e 8  
   - Promotores: notas acima de 8  

3. *Análise exploratória (EDA)*
   - Distribuição do NPS
   - Relação entre atraso e satisfação
   - Impacto do tempo de entrega
   - Análise de contato com atendimento
   - Comparação por região

---

## Principais insights

- A maior parte dos clientes é composta por detratores, indicando baixa satisfação geral
- O atraso na entrega aparece como o principal fator de insatisfação
- Clientes que tiveram mais contato com atendimento tendem a apresentar NPS mais baixo
- O tempo de entrega impacta diretamente a experiência do cliente
- O problema não está concentrado em uma região específica, sugerindo um desafio operacional mais amplo

---

## Conclusão

A análise indica que o principal problema está na experiência pós-compra, especialmente na logística de entrega.

Melhorias na previsibilidade e no tempo de entrega tendem a gerar impacto direto na satisfação dos clientes.
